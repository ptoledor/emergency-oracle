```python
import numpy as np
import pandas as pd
import geopandas as gpd
```

```python
df = gpd.read_file('puntos_emergencias.gpkg', driver='GPKG')
df
```

```python
df = df[df['latitud_x'].notnull()]
df
```

```python
from sklearn.cluster import KMeans
import numpy as np

eventos = df[['utm_x_32718', 'utm_y_32718']].copy()
eventos['evento'] = 'ev_' + eventos.index.astype(str)
eventos.columns = ['ex', 'ey', 'evento']
coords = eventos[['ex', 'ey']].copy().to_numpy()
dmax = 120
stat = []

for k in range(100, 10_000+1):
    model = KMeans(n_clusters=k, random_state=42)
    labels = model.fit_predict(coords)
    ev = eventos.copy()
    ev['k'] = 'k_' + labels.astype(str)
    cides = pd.DataFrame(model.cluster_centers_, columns=['cx', 'cy'])
    cides['k'] = 'k_' + cides.index.astype(str)
    ctoe = pd.merge(cides, ev, on='k', how='inner')
    ctoe['dist'] = np.sqrt((ctoe['cx'] - ctoe['ex'])**2 + (ctoe['cy'] - ctoe['ey'])**2)
    max_dist = ctoe['dist'].max()

    ctoe['ev_in_k'] = ctoe.groupby('k')['evento'].transform('count')

    print(f'k={k}, max_dist={max_dist}, ev_in_k={ctoe["ev_in_k"].max()}, {ctoe["ev_in_k"].mean()}')
    stat.append({'k': k, 'max_dist': max_dist, 'ev_in_k_max': ctoe['ev_in_k'].max(), 'ev_in_k_mean': ctoe['ev_in_k'].mean()})

    if max_dist < dmax:
        break
    
```
